```python
# fibonacci number
n=10
for i in range(n):
    if i <= 1:
        print(i)
    else:
        a, b = 0, 1
        for j in range(2, i + 1):
            c = a + b
            a = b
            b = c   
        print(b)
```

```python
def count_even(numbers):
    count_even = 0

    for i in numbers:
        if i % 2 == 0:
            count_even += 1

    return count_even
```

```python
def remove_duplicates(numbers):
    unique = []

    for i in numbers:
        if i not in unique:
            unique.append(i)

    return unique
```
