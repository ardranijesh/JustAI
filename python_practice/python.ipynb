```python
# fibonacci number
n=10
for i in range(n):
    if i <= 1:
        print(i)
    else:
        a, b = 0, 1
        for j in range(2, i + 1):
            c = a + b
            a = b
            b = c   
        print(b)
```

```python
def count_even(numbers):
    count_even = 0

    for i in numbers:
        if i % 2 == 0:
            count_even += 1

    return count_even
```

```python
def remove_duplicates(numbers):
    unique = []

    for i in numbers:
        if i not in unique:
            unique.append(i)

    return unique
```

```python
def second_largest(numbers):
    f_largest = numbers[0]
    s_largest = numbers[1]

    for i in numbers:
        if i > f_largest:
            f_largest, s_largest = i, f_largest
        elif i > s_largest:
            s_largest = i

    return s_largest
```

```python
def is_palindrome(s):
    for i in range(0, len(s) // 2):
        if s[i] != s[len(s) - 1 - i]:
            return False

    return True
```

```python
# Sum of digits in a number
def sum_digits(n):
    total = 0

    while n > 0:
        digit = n % 10
        total += digit
        n = n // 10

    return total
sum_digits(123)
```

```python
#reverse number
def reverse_number(n):
    rev = 0

    while n > 0:
        x = n % 10
        rev = rev * 10 + x
        n = n // 10

    return rev
```

```python
# pallindrome number
def is_palindrome_number(n):
    rev = 0
    original = n
    while n>0:
        x = n%10
        rev = rev*10 + x
        n = n//10
    return rev == original
is_palindrome_number(121)
# is_palindrome_number(123)
```

```python
# factorial
def factorial(n):
    fact = 1
    while n>1:
        fact= fact*n
        n = n-1
    return fact
factorial(5)
```
