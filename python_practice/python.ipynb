```python
def count_even(numbers):
    count_even = 0

    for i in numbers:
        if i % 2 == 0:
            count_even += 1

    return count_even
```

```python
def remove_duplicates(numbers):
    unique = []

    for i in numbers:
        if i not in unique:
            unique.append(i)

    return unique
```

```python
def second_largest(numbers):
    f_largest = numbers[0]
    s_largest = numbers[1]

    for i in numbers:
        if i > f_largest:
            f_largest, s_largest = i, f_largest
        elif i > s_largest:
            s_largest = i

    return s_largest
```

```python
def is_palindrome(s):
    for i in range(0, len(s) // 2):
        if s[i] != s[len(s) - 1 - i]:
            return False

    return True
```

```python
# Sum of digits in a number
def sum_digits(n):
    total = 0

    while n > 0:
        digit = n % 10
        total += digit
        n = n // 10

    return total
sum_digits(123)
```

```python
#reverse number
def reverse_number(n):
    rev = 0

    while n > 0:
        x = n % 10
        rev = rev * 10 + x
        n = n // 10

    return rev
```

```python
# pallindrome number
def is_palindrome_number(n):
    rev = 0
    original = n
    while n>0:
        x = n%10
        rev = rev*10 + x
        n = n//10
    return rev == original
is_palindrome_number(121)
# is_palindrome_number(123)
```

```python
# factorial
def factorial(n):
    fact = 1
    while n>1:
        fact= fact*n
        n = n-1
    return fact
factorial(5)
```

```python
# fibanocci
def fibanocci(n):
    a,b = 0,1
    fib = []
    for i in range(n):
        fib.append(a)
        a,b = b, a+b
    return fib

fibanocci(6)
# Time coplexity o(n)
# Space complexity o(n)
```

```python
# Prime number
def is_prime(n):
    if n<2 :
        return False
    for i in range(2,n):
        if n%i ==0:
            return False
    return True
print("1:",is_prime(1))
print("2:",is_prime(2))
print("3:",is_prime(3))
print("4:",is_prime(4))
print("5:",is_prime(5))


# Optimised prime
def is_prime_o(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n%i ==0:
            return False
    return True

print("1:",is_prime_o(1))
print("2:",is_prime_o(2))
print("3:",is_prime_o(3))
print("4:",is_prime_o(4))
print("5:",is_prime_o(5))
```

```python
# count frequency
def count_frequency(n):
    frequency = {}
    for i in n:
        if i not in frequency:
            frequency[i] = 1
        else:
            frequency[i] = frequency[i]+1
    return frequency

print("[1, 2, 2, 3, 1, 2, 4] :", count_frequency([1, 2, 2, 3, 1, 2, 4]))
```
